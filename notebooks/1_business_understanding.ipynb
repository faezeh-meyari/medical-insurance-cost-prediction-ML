# Medical Insurance Cost Prediction Project

## Business Understanding

### Project Overview
This project aims to develop a predictive model for estimating medical insurance costs based on various personal and health-related factors. Understanding and predicting medical insurance costs is crucial for both insurance providers and customers, as it helps in better financial planning and risk assessment.

### Business Objectives
1. Develop accurate predictive models for medical insurance cost estimation
2. Identify key factors that influence insurance costs
3. Provide insights for both insurance providers and customers
4. Enable data-driven decision making in insurance pricing



### Project Plan
The project follows the CRISP-DM methodology across 5 notebooks:

1. **Business Understanding (Current Notebook)**
   - Project objectives and success criteria
   - Business context and requirements
   - Project structure and methodology

2. **Data Understanding**
   - Exploratory Data Analysis (EDA)
   - Feature analysis and distributions
   - Correlation analysis
   - No data modifications in this phase

3. **Data Preparation**
   - Data cleaning and preprocessing
   - Feature scaling and encoding
   - Handling missing values and outliers
   - Dataset versioning (raw → interim → processed)

4. **Modeling**
   - Implementation of multiple models:
     - Linear Regression
     - Decision Tree
     - Random Forest
     - Neural Network
   - Model training and hyperparameter tuning

5. **Evaluation**
   - Model comparison and selection
   - Performance metrics analysis
   - Business impact assessment
   - Recommendations and conclusions


## Dataset Overview

The dataset contains information about medical insurance costs and various factors that might influence these costs. Here's a description of the features:

### Features
1. **age**: Age of the primary beneficiary (numeric)
2. **sex**: Gender of the insurance contractor (female/male)
3. **bmi**: Body Mass Index (numeric)
4. **children**: Number of children/dependents covered by insurance (numeric)
5. **smoker**: Smoking status (yes/no)
6. **region**: Beneficiary's residential area in the US (northeast/southeast/southwest/northwest)

### Target Variable
- **charges**: Individual medical costs billed by health insurance (numeric)


### Data Quality Considerations
In the subsequent notebooks, we will analyze:
- Data completeness and missing values
- Distribution of numeric variables
- Categorical variable frequencies
- Outliers and their business significance
- Feature relationships and correlations

### Success Metrics
For our regression models, we will focus on:
1. R² Score (Target > 0.8)
2. Mean Absolute Error (MAE)
3. Root Mean Square Error (RMSE)
